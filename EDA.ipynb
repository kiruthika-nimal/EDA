```python


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


file_path = "global_superstore_2016.xlsx"

df = pd.read_excel(file_path, sheet_name="Orders")

# Convert Order Date to datetime
df["Order Date"] = pd.to_datetime(df["Order Date"])

print("=" * 70)
print("GLOBAL SUPERSTORE SALES ANALYSIS")
print("=" * 70)

# PART 1 – BASIC DATA UNDERSTANDING

# 1. Display first 10 rows
print("\n1. FIRST 10 ROWS")
print("-" * 50)
print(df.head(10))


# 2. Shape, column names and data types
print("\n2. SHAPE OF DATASET")
print("-" * 50)
print("Rows and Columns:", df.shape)

print("\nCOLUMN NAMES:")
print(df.columns.tolist())

print("\nDATA TYPES:")
print(df.dtypes)


# 3. Missing values and duplicates
print("\n3. MISSING VALUES")
print("-" * 50)
print(df.isnull().sum())

print("\nTOTAL MISSING VALUES:", df.isnull().sum().sum())

print("\nDUPLICATE ROWS:", df.duplicated().sum())


# 4. Describe numeric columns
print("\n4. DESCRIPTIVE STATISTICS")
print("-" * 50)
print(df.describe())


# 5. Number of unique customers and products
print("\n5. UNIQUE CUSTOMERS AND PRODUCTS")
print("-" * 50)

unique_customers = df["Customer ID"].nunique()
unique_products = df["Product ID"].nunique()

print("Unique Customers:", unique_customers)
print("Unique Products:", unique_products)


# 6. Country with highest sales
print("\n6. COUNTRY WITH HIGHEST SALES")
print("-" * 50)

country_sales = df.groupby("Country")["Sales"].sum().sort_values(ascending=False)

print(country_sales.head(10))

print("\nCountry with Highest Sales:", country_sales.idxmax())
print("Highest Sales:", country_sales.max())


# PART 2 – EXPLORATORY DATA ANALYSIS

# 1. Histogram of Sales
print("\n" + "=" * 70)
print("PART 2 – EDA")
print("=" * 70)

print("\n1. HISTOGRAM OF SALES")

plt.figure(figsize=(10, 6))
plt.hist(df["Sales"], bins=50)
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.show()


# 2. Top 10 products by Sales
print("\n2. TOP 10 PRODUCTS BY SALES")
print("-" * 50)

top_10_sales = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_10_sales)

plt.figure(figsize=(12, 6))
top_10_sales.sort_values().plot(kind="barh")
plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product Name")
plt.show()


# 3. Top 10 products by Profit
print("\n3. TOP 10 PRODUCTS BY PROFIT")
print("-" * 50)

top_10_profit = (
    df.groupby("Product Name")["Profit"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_10_profit)

plt.figure(figsize=(12, 6))
top_10_profit.sort_values().plot(kind="barh")
plt.title("Top 10 Products by Profit")
plt.xlabel("Total Profit")
plt.ylabel("Product Name")
plt.show()


# 4. Scatter plot Sales vs Profit
print("\n4. SALES VS PROFIT")

plt.figure(figsize=(10, 6))
plt.scatter(df["Sales"], df["Profit"], alpha=0.5)
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()


# 5. Total Sales per Region
print("\n5. TOTAL SALES PER REGION")
print("-" * 50)

region_sales = (
    df.groupby("Region")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print(region_sales)

plt.figure(figsize=(10, 6))
region_sales.plot(kind="bar")
plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()


# 6. Boxplot Profit grouped by Product Category
print("\n6. PROFIT BY PRODUCT CATEGORY")

plt.figure(figsize=(10, 6))
sns.boxplot(data=df, x="Category", y="Profit")
plt.title("Profit Distribution by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Profit")
plt.show()


# 7. Yearly Sales Trend
print("\n7. YEARLY SALES TREND")
print("-" * 50)

df["order_year"] = df["Order Date"].dt.year

yearly_sales = df.groupby("order_year")["Sales"].sum()

print(yearly_sales)

plt.figure(figsize=(10, 6))
plt.plot(yearly_sales.index, yearly_sales.values, marker="o")
plt.title("Yearly Sales Trend")
plt.xlabel("Year")
plt.ylabel("Total Sales")
plt.grid(True)
plt.show()


# PART 3 – STATISTICS

print("\n" + "=" * 70)
print("PART 3 – STATISTICS")
print("=" * 70)


# 1. Mean, Median and Mode of Sales
print("\n1. MEAN, MEDIAN AND MODE OF SALES")
print("-" * 50)

mean_sales = df["Sales"].mean()
median_sales = df["Sales"].median()
mode_sales = df["Sales"].mode()

print("Mean Sales:", mean_sales)
print("Median Sales:", median_sales)
print("Mode Sales:")
print(mode_sales.tolist())


# 2. Variance and Standard Deviation of Profit
print("\n2. VARIANCE AND STANDARD DEVIATION OF PROFIT")
print("-" * 50)

profit_variance = df["Profit"].var()
profit_std = df["Profit"].std()

print("Profit Variance:", profit_variance)
print("Profit Standard Deviation:", profit_std)


# 3. Correlation between Discount and Profit
print("\n3. CORRELATION BETWEEN DISCOUNT AND PROFIT")
print("-" * 50)

correlation = df["Discount"].corr(df["Profit"])

print("Correlation:", correlation)


# 4. Probability Random Order has Discount > 0.3
print("\n4. PROBABILITY OF DISCOUNT > 0.3")
print("-" * 50)

prob_discount = (df["Discount"] > 0.3).mean()

print("Probability:", prob_discount)
print("Percentage:", prob_discount * 100, "%")


# 5. Probability Order is from United States
print("\n5. PROBABILITY OF ORDER FROM UNITED STATES")
print("-" * 50)

prob_usa = (df["Country"] == "United States").mean()

print("Probability:", prob_usa)
print("Percentage:", prob_usa * 100, "%")


# 6. Percentage of orders with negative profit
print("\n6. PERCENTAGE OF ORDERS WITH NEGATIVE PROFIT")
print("-" * 50)

negative_profit_percentage = (df["Profit"] < 0).mean() * 100

print("Negative Profit Orders:", negative_profit_percentage, "%")


# PART 4 – LINEAR ALGEBRA & NUMPY

print("\n" + "=" * 70)
print("PART 4 – LINEAR ALGEBRA & NUMPY")
print("=" * 70)


# 1. Create NumPy arrays
print("\n1. NUMPY ARRAYS")

sales_array = df["Sales"].to_numpy()
profit_array = df["Profit"].to_numpy()

print("Sales Array - First 10:")
print(sales_array[:10])

print("\nProfit Array - First 10:")
print(profit_array[:10])


# 2. Vector Addition and Subtraction
print("\n2. VECTOR ADDITION AND SUBTRACTION")

vector_addition = sales_array + profit_array
vector_subtraction = sales_array - profit_array

print("Addition - First 10:")
print(vector_addition[:10])

print("\nSubtraction - First 10:")
print(vector_subtraction[:10])


# 3. Dot Product
print("\n3. DOT PRODUCT")

dot_product = np.dot(sales_array, profit_array)

print("Dot Product:", dot_product)


# 4. Matrix [Sales, Profit] and weights [0.6, 0.4]
print("\n4. MATRIX MULTIPLICATION")

matrix = np.column_stack((sales_array, profit_array))

weights = np.array([0.6, 0.4])

weighted_result = matrix @ weights

print("Matrix Shape:", matrix.shape)

print("\nWeighted Result - First 10:")
print(weighted_result[:10])


# 5. Normalize Sales Array
print("\n5. NORMALIZED SALES ARRAY")

sales_normalized = (
    (sales_array - sales_array.min()) /
    (sales_array.max() - sales_array.min())
)

print("Normalized Sales - First 10:")
print(sales_normalized[:10])


# PART 5 – CALCULUS

print("\n" + "=" * 70)
print("PART 5 – CALCULUS")
print("=" * 70)


# Profitability Index = (Profit * 100) / Sales

df["Profitability_Index"] = (
    df["Profit"] * 100 / df["Sales"]
)

# Derivative with respect to Profit
# d/dProfit [(Profit * 100) / Sales] = 100 / Sales

df["Derivative_wrt_Profit"] = 100 / df["Sales"]

print("\nProfitability Index and Derivative:")
print(
    df[
        [
            "Sales",
            "Profit",
            "Profitability_Index",
            "Derivative_wrt_Profit"
        ]
    ].head(10)
)

print("\nMathematical Derivative:")
print("d(Profitability Index) / d(Profit) = 100 / Sales")


# PART 6 – FEATURE ENGINEERING

print("\n" + "=" * 70)
print("PART 6 – FEATURE ENGINEERING")
print("=" * 70)


# 1. Profit Margin
df["profit_margin"] = (
    df["Profit"] / df["Sales"]
) * 100

print("\n1. PROFIT MARGIN")
print(df[["Sales", "Profit", "profit_margin"]].head())


# 2. High Discount
df["is_high_discount"] = np.where(
    df["Discount"] > 0.3,
    1,
    0
)

print("\n2. IS HIGH DISCOUNT")
print(df[["Discount", "is_high_discount"]].head())


# 3. Is Profitable
df["is_profitable"] = np.where(
    df["Profit"] > 0,
    1,
    0
)

print("\n3. IS PROFITABLE")
print(df[["Profit", "is_profitable"]].head())


# 4. Sales Quartiles
df["sales_quartile"] = pd.qcut(
    df["Sales"],
    q=4,
    labels=["Q1", "Q2", "Q3", "Q4"]
)

print("\n4. SALES QUARTILES")
print(df[["Sales", "sales_quartile"]].head(10))


# 5. Order Year
df["order_year"] = df["Order Date"].dt.year

print("\n5. ORDER YEAR")
print(df[["Order Date", "order_year"]].head())


# PART 7 – SQL SIMULATION IN PANDAS

print("\n" + "=" * 70)
print("PART 7 – SQL SIMULATION IN PANDAS")
print("=" * 70)


# 1. Profit > 100 and Discount < 0.2
print("\n1. ORDERS WITH PROFIT > 100 AND DISCOUNT < 0.2")

result_1 = df[
    (df["Profit"] > 100) &
    (df["Discount"] < 0.2)
]

print(result_1.head(10))
print("Number of matching orders:", len(result_1))


# 2. Group by Region and calculate total Sales
print("\n2. TOTAL SALES BY REGION")

result_2 = (
    df.groupby("Region")["Sales"]
    .sum()
    .reset_index()
    .sort_values("Sales", ascending=False)
)

print(result_2)


# 3. Top 5 most profitable countries
print("\n3. TOP 5 MOST PROFITABLE COUNTRIES")

result_3 = (
    df.groupby("Country")["Profit"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

print(result_3)


# 4. Orders from 2015 with negative Profit
print("\n4. 2015 ORDERS WITH NEGATIVE PROFIT")

result_4 = df[
    (df["order_year"] == 2015) &
    (df["Profit"] < 0)
]

print(result_4.head(10))
print("Number of matching orders:", len(result_4))


# 5. Top 10 highest sales orders from Corporate segment
print("\n5. TOP 10 HIGHEST SALES CORPORATE ORDERS")

result_5 = (
    df[df["Segment"] == "Corporate"]
    .sort_values("Sales", ascending=False)
    .head(10)
)

print(result_5)


# 6. Count orders in each Shipping Mode
print("\n6. COUNT OF ORDERS BY SHIPPING MODE")

result_6 = df["Ship Mode"].value_counts()

print(result_6)


# PART 8 – INSIGHTS

print("\n" + "=" * 70)
print("PART 8 – INSIGHTS")
print("=" * 70)


# 1. Which country generates the most sales?
country_sales = df.groupby("Country")["Sales"].sum()

highest_sales_country = country_sales.idxmax()
highest_sales_value = country_sales.max()

print("\n1. COUNTRY WITH MOST SALES")
print("Country:", highest_sales_country)
print("Sales:", highest_sales_value)


# 2. Category with highest average profit margin
category_margin = (
    df.groupby("Category")["profit_margin"]
    .mean()
    .sort_values(ascending=False)
)

print("\n2. AVERAGE PROFIT MARGIN BY CATEGORY")
print(category_margin)

print(
    "Highest Average Profit Margin Category:",
    category_margin.idxmax()
)


# 3. Are high discounts always associated with low profits?
high_discount = df[df["Discount"] > 0.3]
low_discount = df[df["Discount"] <= 0.3]

high_discount_profit = high_discount["Profit"].mean()
low_discount_profit = low_discount["Profit"].mean()

print("\n3. HIGH DISCOUNT VS PROFIT")
print("Average Profit - High Discount:",
      high_discount_profit)

print("Average Profit - Normal Discount:",
      low_discount_profit)

print("Discount-Profit Correlation:",
      correlation)

if correlation < 0:
    print("Conclusion: Higher discounts tend to be associated with lower profits.")
elif correlation > 0:
    print("Conclusion: Higher discounts tend to be associated with higher profits.")
else:
    print("Conclusion: There is no linear relationship between discount and profit.")


# Scatter plot
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit"
)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")
plt.show()


# 4. Which region contributes most to sales?
highest_sales_region = region_sales.idxmax()

print("\n4. REGION WITH MOST SALES")
print("Region:", highest_sales_region)
print("Sales:", region_sales.max())


# 5. Do large orders tend to be more profitable?
sales_profit_correlation = df["Sales"].corr(df["Profit"])

print("\n5. SALES VS PROFIT CORRELATION")
print("Correlation:", sales_profit_correlation)

if sales_profit_correlation > 0:
    print("Conclusion: Larger sales generally tend to be associated with higher profit.")
elif sales_profit_correlation < 0:
    print("Conclusion: Larger sales tend to be associated with lower profit.")
else:
    print("Conclusion: No linear relationship between sales and profit.")


# 6. Which year had highest sales?
highest_sales_year = yearly_sales.idxmax()

print("\n6. YEAR WITH HIGHEST SALES")
print("Year:", highest_sales_year)
print("Sales:", yearly_sales.max())


# 7. Percentage of orders resulting in loss
loss_percentage = (
    (df["Profit"] < 0).mean() * 100
)

print("\n7. PERCENTAGE OF ORDERS RESULTING IN LOSS")
print("Loss Percentage:", loss_percentage, "%")


# EXTRA TASKS – ADDITIONAL ANALYSIS

print("\n" + "=" * 70)
print("EXTRA ANALYSIS")
print("=" * 70)


# 1. Most profitable category
category_profit = (
    df.groupby("Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

print("\n1. MOST PROFITABLE CATEGORY")
print(category_profit)
print("Most Profitable Category:",
      category_profit.idxmax())


# 2. Most profitable sub-category
subcategory_profit = (
    df.groupby("Sub-Category")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

print("\n2. MOST PROFITABLE SUB-CATEGORY")
print(subcategory_profit)
print("Most Profitable Sub-Category:",
      subcategory_profit.idxmax())


# 3. Most profitable customer segment
segment_profit = (
    df.groupby("Segment")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

print("\n3. MOST PROFITABLE CUSTOMER SEGMENT")
print(segment_profit)
print("Most Profitable Segment:",
      segment_profit.idxmax())


# 4. Sales and Profit by Year
year_analysis = (
    df.groupby("order_year")[["Sales", "Profit"]]
    .sum()
)

print("\n4. SALES AND PROFIT BY YEAR")
print(year_analysis)


# 5. Profit Margin by Category
category_profit_margin = (
    df.groupby("Category")["profit_margin"]
    .mean()
    .sort_values(ascending=False)
)

print("\n5. PROFIT MARGIN BY CATEGORY")
print(category_profit_margin)

plt.figure(figsize=(10, 6))

category_profit_margin.plot(kind="bar")

plt.title("Average Profit Margin by Category")
plt.xlabel("Category")
plt.ylabel("Average Profit Margin (%)")

plt.xticks(rotation=0)

plt.show()


# 6. Top 10 Customers by Sales
top_customers = (
    df.groupby("Customer Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("\n6. TOP 10 CUSTOMERS BY SALES")
print(top_customers)


# 7. Top 10 Customers by Profit
top_profit_customers = (
    df.groupby("Customer Name")["Profit"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("\n7. TOP 10 CUSTOMERS BY PROFIT")
print(top_profit_customers)


# 8. Overall Sales, Profit and Profit Margin
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()

overall_profit_margin = (
    total_profit / total_sales
) * 100

print("\n8. OVERALL BUSINESS PERFORMANCE")
print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
print("Overall Profit Margin:", overall_profit_margin, "%")


# 9. High Discount vs Normal Discount Analysis
discount_analysis = df.groupby(
    "is_high_discount"
).agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Average_Profit=("Profit", "mean"),
    Average_Discount=("Discount", "mean")
)

print("\n9. HIGH DISCOUNT ANALYSIS")
print(discount_analysis)


# 10. Profitable vs Loss-Making Orders
profitability_count = df["is_profitable"].value_counts()

print("\n10. PROFITABLE VS LOSS-MAKING ORDERS")
print(profitability_count)

plt.figure(figsize=(8, 5))

profitability_count.plot(kind="bar")

plt.title("Profitable vs Loss-Making Orders")
plt.xlabel("Is Profitable (1 = Yes, 0 = No)")
plt.ylabel("Number of Orders")

plt.xticks(rotation=0)

plt.show()


# FINAL SUMMARY

print("\n" + "=" * 70)
print("FINAL SUMMARY")
print("=" * 70)

print("\n1. Country with highest sales:")
print(highest_sales_country)

print("\n2. Region with highest sales:")
print(highest_sales_region)

print("\n3. Year with highest sales:")
print(highest_sales_year)

print("\n4. Most profitable category:")
print(category_profit.idxmax())

print("\n5. Most profitable sub-category:")
print(subcategory_profit.idxmax())

print("\n6. Most profitable customer segment:")
print(segment_profit.idxmax())

print("\n7. Total Sales:")
print(total_sales)

print("\n8. Total Profit:")
print(total_profit)

print("\n9. Overall Profit Margin:")
print(overall_profit_margin, "%")

print("\n10. Loss-making orders:")
print(loss_percentage, "%")

print("\n11. Discount-Profit Correlation:")
print(correlation)

print("\n12. Unique Customers:")
print(unique_customers)

print("\n13. Unique Products:")
print(unique_products)

print("\n" + "=" * 70)
print("ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 70)
```
